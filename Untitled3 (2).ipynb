```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score

try:
    df = pd.read_csv('/content/fake reviews dataset.csv')
    print(f"Dataset loaded successfully. Shape: {df.shape}")
except FileNotFoundError:
    print("Error: Could not find 'fake reviews dataset.csv'. Please ensure it is in the correct folder.")

print(f"DataFrame columns after loading: {df.columns.tolist()}")

if 'label' in df.columns:
    df['binary_label'] = df['label'].map({'CG': 1, 'OR': 0})
else:
    print("Warning: 'label' column not found. Check your CSV column names.")

df.dropna(subset=['text_', 'binary_label'], inplace=True)

X = df['text_']
y = df['binary_label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)


pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english', max_features=5000, ngram_range=(1, 2))),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1))
])

print("Training the Random Forest model (this may take a moment on a large dataset)...")
pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)

print("\nModel Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=["Real (0)", "Fake (1)"]))

new_reviews = [
    "I absolutely love this! Best purchase of my life, everyone buy this right now!",
    "The product arrived scratched. It works, but the packaging was damaged."
]

predictions = pipeline.predict(new_reviews)

print("\nTesting new reviews:")
for review, pred in zip(new_reviews, predictions):
    label = "Fake" if pred == 1 else "Real"
    print(f"- [{label}] {review}")
```

Feature Engineering (Adding new columns)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

df = pd.read_csv('/content/fake reviews dataset.csv')
df.dropna(subset=['text_', 'label', 'rating'], inplace=True)

df['binary_label'] = df['label'].map({'CG': 1, 'OR': 0})

df['review_length'] = df['text_'].apply(lambda x: len(str(x)))

X = df[['text_', 'rating', 'review_length']]
y = df['binary_label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

The ColumnTransformer & Pipeline

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('text_tfidf', TfidfVectorizer(stop_words='english', max_features=5000, ngram_range=(1, 2)), 'text_'),
        ('num_scaler', StandardScaler(), ['rating', 'review_length'])
    ],
    remainder='drop'
)

pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1))
])

print("Training the enhanced Random Forest model...")
pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)

print("\nEnhanced Model Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print(classification_report(y_test, y_pred, target_names=["Real (0)", "Fake (1)"]))
```

Building the Comparison Table (XGBoost & Naive Bayes)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

df = pd.read_csv('/content/fake reviews dataset.csv')
df.dropna(subset=['text_', 'label', 'rating'], inplace=True)
df['binary_label'] = df['label'].map({'CG': 1, 'OR': 0})
df['review_length'] = df['text_'].apply(lambda x: len(str(x)))

X = df[['text_', 'rating', 'review_length']]
y = df['binary_label']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

preprocessor = ColumnTransformer(
    transformers=[
        ('text_tfidf', TfidfVectorizer(stop_words='english', max_features=5000, ngram_range=(1, 2)), 'text_'),
        ('num_scaler', MinMaxScaler(), ['rating', 'review_length'])
    ],
    remainder='drop'
)

models = {
    "Naive Bayes": MultinomialNB(),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    "XGBoost": XGBClassifier(n_estimators=100, learning_rate=0.1, random_state=42, use_label_encoder=False, eval_metric='logloss')
}

results = []

print("Starting Model Benchmarking...\n")

for model_name, model in models.items():
    print(f"Training {model_name}...")

    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, pos_label=1)
    rec = recall_score(y_test, y_pred, pos_label=1)
    f1 = f1_score(y_test, y_pred, pos_label=1)

    results.append({
        "Model": model_name,
        "Accuracy": f"{acc * 100:.2f}%",
        "Precision (Fake)": f"{prec:.2f}",
        "Recall (Fake)": f"{rec:.2f}",
        "F1-Score (Fake)": f"{f1:.2f}"
    })

print("\n--- Final Research Paper Benchmark Table ---")
results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```

Feature Importance (The Final Piece of the Puzzle)

```python
import matplotlib.pyplot as plt
import numpy as np

print("Generating Feature Importance Chart for Random Forest...")

rf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1))
])
rf_pipeline.fit(X_train, y_train)

tfidf_features = rf_pipeline.named_steps['preprocessor'].transformers_[0][1].get_feature_names_out()
num_features = ['rating', 'review_length']
all_features = np.concatenate([tfidf_features, num_features])

importances = rf_pipeline.named_steps['classifier'].feature_importances_

importance_df = pd.DataFrame({'Feature': all_features, 'Importance': importances})
top_20_features = importance_df.sort_values(by='Importance', ascending=False).head(20)

plt.figure(figsize=(10, 8))
plt.barh(top_20_features['Feature'][::-1], top_20_features['Importance'][::-1], color='steelblue')
plt.title('Top 20 Most Important Features for Fake Review Detection')
plt.xlabel('Relative Importance (Gini Importance)')
plt.ylabel('Feature')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.savefig('feature_importance.png')
plt.show()
```

```python
import shap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Initializing Explainable AI (SHAP) Analysis...")

preprocessor = pipeline.named_steps['preprocessor']
rf_model = pipeline.named_steps['classifier']

X_test_sample = X_test.sample(n=500, random_state=42)
X_test_transformed = preprocessor.transform(X_test_sample)

all_feature_names = preprocessor.get_feature_names_out()

X_test_dense = pd.DataFrame(X_test_transformed.toarray(), columns=all_feature_names)

explainer = shap.TreeExplainer(rf_model)

raw_shap_values = explainer.shap_values(X_test_dense, check_additivity=False)

if isinstance(raw_shap_values, list):
    shap_values = raw_shap_values[1]
elif isinstance(raw_shap_values, np.ndarray) and raw_shap_values.ndim == 2:
    shap_values = raw_shap_values
elif isinstance(raw_shap_values, np.ndarray) and raw_shap_values.ndim == 3:
    shap_values = raw_shap_values[:, :, 1]
else:
    print(f"Warning: SHAP values returned in an unexpected format: type={type(raw_shap_values)}, ndim={getattr(raw_shap_values, 'ndim', 'N/A')}")
    shap_values = raw_shap_values

plt.figure(figsize=(12, 8))
plt.title("SHAP Summary: Global Feature Impact on Fake Review Detection")
shap.summary_plot(shap_values, X_test_dense, show=False)
plt.tight_layout()
plt.savefig('shap_summary_plot.png', dpi=300, bbox_inches='tight')
plt.show()

sample_index = 0
print(f"\nAnalyzing specific review text: '{X_test_sample.iloc[sample_index]}'")

shap.initjs()

expected_val = explainer.expected_value[1] if isinstance(explainer.expected_value, (list, np.ndarray)) else explainer.expected_value

force_plot = shap.force_plot(
    expected_val,
    shap_values[sample_index],
    X_test_dense.iloc[sample_index],
    matplotlib=True,
    show=False
)
plt.savefig('shap_single_prediction.png', dpi=300, bbox_inches='tight')
plt.show()
```
